```python
# this function moves a card to the right and updates review date
def move_card_right(df, card_id):
    if df.at[card_id, "Box"] < 5:
        df.at[card_id, "Box"] += 1
        update_review_dates(df, card_id)
        df.to_excel(sheet)

# this function moves a card to the left and updates review date
def move_card_left(df, card_id):
    if df.at[card_id, "Box"] > 1:
        df.at[card_id, "Box"] -= 1
        update_review_dates(df, card_id)
        df.to_excel(sheet)
    elif df.at[card_id, "Box"] == 1:
        update_review_dates(df, card_id)
        df.to_excel(sheet)

#this functions takes the box number and returns the date of next review
def calculate_next_review(box):
    today = datetime.today().replace(microsecond=0)
    if box == 1:
        return today + timedelta(days=1)
    elif box == 2:
        return today + timedelta(days=2)
    elif box == 3:
        return today + timedelta(weeks=1)
    elif box == 4:
        return today + timedelta(weeks=2)
    elif box == 5:
        return today + timedelta(weeks=4)
    else:
        return today

#this function uses the calculated review date updates updates the dataFrame
def update_review_dates(df, card_id):
    today = datetime.today().replace(microsecond=0)  # remove microseconds
    df.at[card_id, "Last Review"] = today
    current_box = df.at[card_id, "Box"]
    df.at[card_id, "Next Review"] = calculate_next_review(current_box)

def draw_card():
    now = datetime.today().replace(microsecond=0)
    # check if there are more cards left
    while True:
        due_cards = df[
            (df["Next Review"].isna()) |
            (df["Next Review"] <= now)
        ]

        if due_cards.empty:
            stackStatus = "empty"
            break

        #choose a random index
        idx = random.choice(due_cards.index.tolist())
        #choose a random card
        return df.loc[idx, "Question"]


        #while True:
        #    user_input = input("Was it correct or incorrect? (r/f): ").lower()
        #    if user_input == "r":
        #        move_card_right(df, idx)
        #        break
        #    elif user_input == "f":
        #        move_card_left(df, idx)
        #        break
        #    else:
        #        print("Invalid input, type 'r' for right or 'f' for fail.")


```
